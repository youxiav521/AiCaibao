```python
from fastmcp import Client

client = Client("http://127.0.0.1:6300/api/coding_agent_mcp/mcp")

async def main():
    async with client:
        ping_res = await client.ping()
        print("ping结果:", ping_res)
        tools = await client.list_tools()
        print("工具列表:", tools)

        keepalive_count = 0

        async def progress_handler(event,*args, **kwargs):
            print(f"event:{event}")
            print(f"args[1]:{args[1]}")
            nonlocal keepalive_count
            if  "keepalive" in str(args[1]):
                keepalive_count += 1

        # 发送请求并以流式方式接收响应，检测 keepalive 机制
        r = await client.call_tool(
            "coding_agent",
            {
                "question": "请根据文档，回答问题：2024谁的年薪最高？",
                "prompt": "注意理解给出的数据，可能是表格数据或者是文本数据，需要根据数据进行分析，并理解用户的需要 /no_think",
                "documents": [
                    {
                        "id": 1,
                        "title": "2025年9月8日 table-1",
                        "text": "",
                        "table": "姓名,年龄,城市,职业,薪资,年发薪次数\n张三,28,北京,软件工程师,15000,10\n李四,32,上海,产品经理,18000,12\n王五,25,深圳,设计师,12000,8\n赵六,30,杭州,数据分析师,16000,15"
                    },
                    {
                        "id": 2,
                        "title": "2024年9月8日 table-2",
                        "text": "",
                        "table": "姓名,年龄,城市,职业,薪资,年发薪次数\n张三,28,北京,软件工程师,15000,10\n李四,32,上海,产品经理,18000,15\n王五,25,深圳,设计师,12000,8\n赵六,30,杭州,数据分析师,16000,15"
                    },
                    {
                        "id":3,
                        "title": "公司晋升指标",
                        "text": "本公司的晋升指标是：1. 公司指标1 2. 公司指标2 3. 公司指标3",
                        "table": ""
                    },
                    {
                        "id":4,
                        "title": "公司XX指标",
                        "text": "本公司的XX指标是：1. 公司指标1 2. 公司指标2 3. 公司指标3",
                        "table": ""
                    },
                    {
                        "id":5,
                        "title": "公司XX指标",
                        "text": "本公司的XX指标是：1. 公司指标1 2. 公司指标2 3. 公司指标3",
                        "table": ""
                    }
                ]
            },
            timeout=3600,
            progress_handler=progress_handler
        )
        # r 为最终响应
        print("最终响应:", r)
        if keepalive_count > 0:
            print(f"keepalive 机制正常，共收到 {keepalive_count} 次 keepalive 推送。")
        else:
            print("未检测到 keepalive 推送，请检查服务端实现。")

# 启动主协程
await main()
```
