```python
# 快速测试大模型响应
import sys
import os
from pathlib import Path

# 从当前目录向上查找包含 src 的目录作为项目根（兼容 tests/notebooks 或其它子目录运行）
_candidate = Path().absolute()
for _ in range(6):
    if (_candidate / "src").is_dir():
        project_root = _candidate
        break
    if _candidate.parent == _candidate:
        project_root = Path().absolute().parent.parent  # 回退：notebooks -> tests -> 项目根
        break
    _candidate = _candidate.parent
else:
    project_root = _candidate

if str(project_root) not in sys.path:
    sys.path.insert(0, str(project_root))
os.chdir(project_root)
print(f"📂 项目根: {project_root}")
print(f"📂 工作目录: {os.getcwd()}")

import asyncio
import time
from datetime import datetime
from src.utils.model_factory import ChatModelFactory

async def quick_ping_model(model_id: str = "model_for_content_generation", timeout: int = 10):
    """
    快速ping测试大模型响应
    
    Args:
        model_id: 模型ID
        timeout: 超时时间（秒）
    """
    print(f"🚀 开始测试模型: {model_id}")
    start_time = time.time()
    
    try:
        # 创建模型（配置路径已经修正）
        factory = ChatModelFactory()
        model = factory.create_langchain_model(model_id)

        # 发送测试消息
        print("📤 发送测试消息: '你好'")
        test_message = """你好
        """
        
        response = await asyncio.wait_for(
            model.ainvoke([{"role": "user", "content": test_message}]),
            timeout=timeout
        )
        
        end_time = time.time()
        elapsed_time = end_time - start_time
        
        print(f"✅ 测试成功！")
        print(f"⏱️  响应时间: {elapsed_time:.2f}秒")
        print(f"📝 模型响应: {response.content[:200] if hasattr(response, 'content') else str(response)[:200]}")
        
        return True
        
    except asyncio.TimeoutError:
        print(f"❌ 请求超时（>{timeout}秒）")
        return False
    except Exception as e:
        print(f"❌ 发生错误: {e}")
        import traceback
        traceback.print_exc()
        return False

# 测试主模型
result = await quick_ping_model("model_bailian_30b", timeout=15)
print(f"\n测试结果: {'成功 ✅' if result else '失败 ❌'}")
```
